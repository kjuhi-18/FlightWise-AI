```python
# Columns to remove

cols_to_drop = [

    # Leakage columns
    'DepTime',
    'DepDelay',
    'DepDelayMinutes',
    'DepDel15',
    'DepartureDelayGroups',

    'ArrTime',
    'ArrDelay',
    'ArrDelayMinutes',
    'ArrivalDelayGroups',

    'CarrierDelay',
    'WeatherDelay',
    'NASDelay',
    'SecurityDelay',
    'LateAircraftDelay',

    # Post-flight columns
    'TaxiOut',
    'TaxiIn',
    'WheelsOff',
    'WheelsOn',
    'AirTime',
    'ActualElapsedTime',

    # DOT IDs
    'DOT_ID_Marketing_Airline',
    'DOT_ID_Originally_Scheduled_Code_Share_Airline',
    'DOT_ID_Operating_Airline',

    # Airport IDs
    'OriginAirportID',
    'OriginAirportSeqID',
    'OriginCityMarketID',

    'DestAirportID',
    'DestAirportSeqID',
    'DestCityMarketID',

    # Geographic codes
    'OriginStateFips',
    'DestStateFips',
    'OriginWac',
    'DestWac',

    # Misc
    'Tail_Number',
    'Duplicate',
    'Flights',
    'Unnamed: 119'
]

# Drop all Div1–Div5 columns automatically
div_cols = [col for col in df.columns if col.startswith('Div')]

# Combine
cols_to_drop.extend(div_cols)

# IMPORTANT:
# Keep ArrDel15 as target, so don't drop it
if 'ArrDel15' in cols_to_drop:
    cols_to_drop.remove('ArrDel15')

# Drop columns
df = df.drop(columns=cols_to_drop, errors='ignore')

print("Remaining columns:", len(df.columns))
print(df.columns.tolist())
```

```python
# Additional columns to remove

cols_to_drop = [
    # Redundant airline codes
    'IATA_Code_Marketing_Airline',
    'IATA_Code_Operating_Airline',
    'IATA_Code_Originally_Scheduled_Code_Share_Airline',

    # Redundant location columns
    'OriginCityName',
    'DestCityName',
    'OriginState',
    'DestState',

    # Post-operation information (data leakage)
    'FirstDepTime',
    'TotalAddGTime',
    'LongestAddGTime'
]

# Drop columns
df.drop(columns=cols_to_drop, inplace=True, errors='ignore')

print(f"Remaining columns: {len(df.columns)}")
print(df.columns.tolist())
```

```python
df.drop(columns=['FlightDate'], inplace=True)
```

```python
df.drop(columns=[
    'Flight_Number_Marketing_Airline',
    'Flight_Num_Originally_Scheduled_Code_Share_Airline',
    'Flight_Number_Operating_Airline'
], inplace=True)
```

```python
df.drop(
    columns=['Cancelled', 'CancellationCode'],
    inplace=True
)
```
